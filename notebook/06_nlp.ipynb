```python
import re
import pandas as pd
import numpy as np
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

import warnings
warnings.filterwarnings("ignore")
```

```python
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
feedback = [
    "I am struggling with mathematics and need more practice.",
    "The machine learning classes are very interesting.",
    "I cannot understand Python programming properly.",
    "I enjoy learning artificial intelligence.",
    "The teacher explains difficult topics very clearly.",
    "I need help with deep learning concepts.",
    "My performance in mathematics is improving.",
    "I find programming difficult and confusing."
]

df = pd.DataFrame({
    "feedback": feedback
})

df
```

```python
df["lowercase"] = df["feedback"].str.lower()

df[["feedback", "lowercase"]]
```

```python
def clean_text(text):

    text = re.sub(r"<.*?>", "", text)

    text = re.sub(r"http\S+|www\S+", "", text)

    text = re.sub(r"[^a-zA-Z\s]", "", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()


df["cleaned_text"] = df["lowercase"].apply(clean_text)

df[["feedback", "cleaned_text"]]
```

```python
df["tokens"] = df["cleaned_text"].apply(lambda x: x.split())

df[["cleaned_text", "tokens"]]
```

```python
stop_words = set(stopwords.words("english"))

def remove_stopwords(words):

    return [word for word in words if word not in stop_words]


df["without_stopwords"] = df["tokens"].apply(remove_stopwords)

df[["tokens", "without_stopwords"]]
```

```python
stemmer = PorterStemmer()

def stemming(words):

    return [stemmer.stem(word) for word in words]


df["stemmed"] = df["without_stopwords"].apply(stemming)

df[["without_stopwords", "stemmed"]]
```

```python
lemmatizer = WordNetLemmatizer()

def lemmatization(words):

    return [lemmatizer.lemmatize(word) for word in words]


df["lemmatized"] = df["without_stopwords"].apply(lemmatization)

df[["without_stopwords", "lemmatized"]]
```

```python
corpus = df["cleaned_text"]

bow = CountVectorizer()

bow_matrix = bow.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow.get_feature_names_out()
)

bow_df
```

```python
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(corpus)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df
```

```python
word_importance = tfidf_df.mean().sort_values(ascending=False)

print("Top 10 important words:")

display(word_importance.head(10))
```
